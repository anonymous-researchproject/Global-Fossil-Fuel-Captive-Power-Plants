```python
import rioxarray as rxr
import numpy as np
import rasterio
from rasterio.warp import reproject, Resampling
from scipy.ndimage import distance_transform_edt
```

# Preprocessing for four main agricultural crops; 2020 literature yield data derived from 2015 literature data (yield data required to calculate residue)

```python
# --------------------------
# Input files
# --------------------------
ggcp_path = r"./data/biomass_supply/agricultural_production_map_4main/GGCP10_Production_2020_Maize.tif"
gaez_path = r"./data/biomass_supply/agricultural yield/GAEZ_2015_crop_yield/GAEZAct2015_Yield_Maize_Mean.tif"
output_path = r"./data/biomass_supply/agricultural_production_map_4main/Maize_yield_tonnes_per_ha.tif"

# --------------------------
# Read GGCP
# --------------------------
with rasterio.open(ggcp_path) as ggcp_src:
    ggcp = ggcp_src.read(1)
    meta = ggcp_src.meta.copy()
    transform = ggcp_src.transform
    crs = ggcp_src.crs

# --------------------------
# Read GAEZ
# --------------------------
with rasterio.open(gaez_path) as gaez_src:
    gaez = gaez_src.read(1)

    # If grids differ, resample to GGCP
    if (gaez_src.crs != crs) or (gaez_src.transform != transform) or (gaez_src.shape != ggcp.shape):

        gaez_resampled = np.empty_like(ggcp, dtype=np.float32)

        reproject(
            source=gaez,
            destination=gaez_resampled,
            src_transform=gaez_src.transform,
            src_crs=gaez_src.crs,
            dst_transform=transform,
            dst_crs=crs,
            resampling=Resampling.nearest
        )

        gaez = gaez_resampled

# --------------------------
# Create mask
# --------------------------
ggcp_mask = ggcp > 0
valid_mask = gaez > 0

if not np.any(valid_mask):
    raise ValueError("GAEZ has no cells greater than 0")

# --------------------------
# Distance transform (core)
# --------------------------
distance, indices = distance_transform_edt(
    ~valid_mask,
    return_indices=True
)

# indices have shape=(2, rows, cols)
nearest_rows = indices[0]
nearest_cols = indices[1]

# --------------------------
# Create output
# --------------------------
output = np.zeros_like(gaez, dtype=np.float32)

# Assign values only where GGCP > 0
output[ggcp_mask] = gaez[
    nearest_rows[ggcp_mask],
    nearest_cols[ggcp_mask]
]

# --------------------------
# Write output
# --------------------------
meta.update(dtype='float32', nodata=0)

with rasterio.open(output_path, 'w', **meta) as dst:
    dst.write(output, 1)

print("Done: Maize_yield_tonnes_per_ha.tif generated")
```

# Main calculation: agricultural residue

### Maize    tons per 25km2

```python
# =====================================================
# 1️⃣ Read input rasters (masked=True auto-handles nodata)
# =====================================================
file_A = "./data/biomass_supply/agricultural_production_map_4main/Maize_yield_tonnes_per_ha.tif"
file_B = "./data/biomass_supply/agricultural_production_map_4main/GGCP10_Production_2020_Maize.tif"

rds_A = rxr.open_rasterio(file_A, masked=True).squeeze()
rds_B = rxr.open_rasterio(file_B, masked=True).squeeze()

# =====================================================
# 2️⃣ Apply your formula on file A
# =====================================================
yield_data = rds_A.where(rds_A > 0)
ratio_data = -0.138 * np.log(yield_data * 100) + 1.8681

# =====================================================
# 3️⃣ Multiply the result with file B
# =====================================================
residue_data = ratio_data * rds_B * 1000 / 4      # *1000 converts unit from 1000 tons to tons; dividing by 4 converts raster from 10 km to 5 km

# =====================================================
# 4️⃣ Restore nodata (use A's nodata)
# =====================================================
nodata_value = rds_A.rio.nodata
residue_data = residue_data.fillna(nodata_value)

# =====================================================
# 5️⃣ Resample to 5 km (≈0.0416667°) using nearest neighbor
# =====================================================
target_resolution = 0.04166666666667 if residue_data.rio.crs.is_geographic else 5000

residue_data_5km = residue_data.rio.reproject(
    residue_data.rio.crs,
    resolution=target_resolution,
    resampling=Resampling.nearest
)

# =====================================================
# 6️⃣ Save the resampled raster
# =====================================================
output_tif = "./data/biomass_supply/agricultural_residue_map/maize_residue.tif"
residue_data_5km.rio.to_raster(output_tif, nodata=nodata_value) #unit is tons per 25km2

print("Done: Multiplication and resampling complete. Saved as:", output_tif)
```

### Soybean: directly uses residue-to-production ratio, relative to yield, so no conversion needed

```python
# =====================================================
# Read original tif
# =====================================================
input_tif = r"./data/Crop residue management dataset(1997_2021)/Global Gridded Crop Production Dataset at 10 km Resolution from 2010 to 2020/GGCP10_Production_2020_Soybean.tif"
rds = rxr.open_rasterio(input_tif)

# =====================================================
# Specify target resolution (unit must match CRS)
# =====================================================
target_resolution = 0.04166666666667 if rds.rio.crs.is_geographic else 5000

# =====================================================
# Resample
# =====================================================
rds_resampled = rds.rio.reproject(
    rds.rio.crs,
    resolution=target_resolution,
    resampling=Resampling.nearest
)

# =====================================================
# Linear transformation: divide by 4 then multiply by 1000
# =====================================================
rds_transformed = (rds_resampled / 4) * 1000

# =====================================================
# Create mask
#     Original invalid values or values <= 0 are treated as mask
# =====================================================
mask = (rds_resampled == -9999) | (rds_transformed <= 0)

# =====================================================
# Safe formula calculation
# =====================================================
# Mask regions first to avoid log(0) or negative values
yield_data = rds_transformed.where(~mask)

# Residue formula calculation
residue_data = 2.370 * yield_data

# =====================================================
# Restore masked regions to 0
# =====================================================
residue_data = residue_data.where(~mask, 0)

# =====================================================
# Save as new tif, set nodata = 0
# =====================================================
output_tif = r"./data/biomass_supply/agricultural_residue_map/soybean_residue.tif"
residue_data.rio.to_raster(output_tif, nodata=0)  #tons per 25km2

print("Done: resampling, linear transformation and residue calculation complete:", output_tif)
```

### Calculate total potential

```python
import rasterio
from rasterio.warp import reproject, Resampling
from rasterio.transform import from_origin
import numpy as np
import glob
import os
import math
```

```python
# ---------- Configuration ----------
input_folder = r"./data/biomass_supply/agricultural_theoretical_energy/initial"
output_tif = os.path.join(input_folder, "overall_biomass_ag_half_SCR_global.tif")

SCR = 0.5 #Sustainable Collection Rate
resolution_deg = 0.04166666666667  # ≈ 5 km at equator
crs_target = "EPSG:4326"
# --------------------------

# Calculate global extent (-180, -90, 180, 90)
minx, miny, maxx, maxy = -180, -90, 180, 90
width = int(math.ceil((maxx - minx) / resolution_deg))
height = int(math.ceil((maxy - miny) / resolution_deg))
transform = from_origin(minx, maxy, resolution_deg, resolution_deg)

print(f"Creating global grid: {width} x {height}, CRS={crs_target}")

# Initialize accumulation array
accum = np.zeros((height, width), dtype=np.float64)

# Get all tif files
tif_files = sorted(glob.glob(os.path.join(input_folder, "*.tif")))
print(f"Found {len(tif_files)} tif files")

# Helper function: set nodata, negative values, and NaN to 0
def preprocess_data(data, nodata_value):
    if nodata_value is not None:
        data[data == nodata_value] = 0
    data[data < 0] = 0
    data = np.nan_to_num(data, nan=0.0)
    return data

# Loop through each raster
for f in tif_files:
    with rasterio.open(f) as src:
        print("Processing:", os.path.basename(f))
        data = src.read(1).astype(np.float64)
        data = preprocess_data(data, src.nodata)

        # Create destination array
        dst = np.zeros((height, width), dtype=np.float64)

        # Reproject to global grid
        reproject(
            source=data,
            destination=dst,
            src_transform=src.transform,
            src_crs=src.crs,
            dst_transform=transform,
            dst_crs=crs_target,
            resampling=Resampling.nearest,
            src_nodata=0,
            dst_nodata=0,
        )

        # Accumulate
        accum += dst

# Ensure accumulated result has no NaN
accum = np.nan_to_num(accum, nan=0.0)

# Multiply by SCR coefficient
accum *= SCR

# Output GeoTIFF
meta = {
    "driver": "GTiff",
    "height": height,
    "width": width,
    "count": 1,
    "dtype": "float32",
    "crs": crs_target,
    "transform": transform,
    "nodata": 0.0,
    "compress": "LZW"
}

with rasterio.open(output_tif, "w", **meta) as dst:
    dst.write(accum.astype("float32"), 1)

print("Done: output saved to:", output_tif)
```
